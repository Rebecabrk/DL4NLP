Consider:
- **S1: The doctor examined the child.**
- **S2: The physician examined the kid.**
- **S3: The child examined the doctor.**

a) Build a set representation and a bag-of-words vector representation for the three sentences.

b) Compute Jaccard similarity and cosine similarity for S1-S2 and S1-S3. Show the representations

c) Compute a token-level edit distance for the same pairs.

d) Using WordNet, inspect the relations/similarity between doctor-physician and child-kid. Specify
the synsets used.

e) Which pair is semantically closer? Which representation(s) capture this correctly? Explain why
S1 and S3 may appear deceptively similar in a bag-of-words representation.

## d)
* "Doctor" and "physician" are synonyms because they belong to the same synset (with the same meaning as in the provided sentences).

(noun) **doctor**, doc, **physician**, MD, Dr., medico: *a licensed medical practitioner*

    "I felt so bad I went to see my doctor"

[https://en-word.net/view/lemma/doctor]
[https://en-word.net/view/lemma/physician]

common synset for "doctor" and "physician": doctor.n.01 -- a licensed medical practitioner


* "Child" and "kid" are synonyms because they belong to the same synsets (with the same meaning as in the provided sentences).

(noun) **child**, **kid**, youngster, minor, shaver, nipper, small fry, tiddler, tike, tyke, fry, nestling: *a young person of either sex* (synset child.n.01)

    "she writes books for children"
    "they're just kids"
    "'tiddler' is a British term for youngster"

(noun) **child**, **kid**: *a human offspring (son or daughter) of any age* (synset child.n.02)

    "they had three children"
    "they were able to send their kids to college"

[https://en-word.net/view/lemma/child]
[https://en-word.net/view/lemma/kid]

```python
from nltk.corpus import wordnet as wn

print(wn.synsets('doctor'))
# output: [Synset('doctor.n.01'), Synset('doctor_of_the_church.n.01'), Synset('doctor.n.03'), Synset('doctor.n.04'), Synset('sophisticate.v.03'), Synset('doctor.v.02'), Synset('repair.v.01')]

print(wn.synsets('doctor', pos=wn.NOUN)) # POS = PART OF SPEECH
# output: [Synset('doctor.n.01'), Synset('doctor_of_the_church.n.01'), Synset('doctor.n.03'), Synset('doctor.n.04')]

print(wn.synset('doctor.n.01').definition())
# output: a licensed medical practitioner
print(wn.synset('doctor.n.02').definition())
# output: (Roman Catholic Church) a title conferred on 33 saints who distinguished themselves through the orthodoxy of their theological teaching
print(wn.synset('doctor.n.03').definition())
# output: children take the roles of physician or patient or nurse and pretend they are at the physician's office
print(wn.synset('doctor.n.04').definition())
# output: a person who holds Ph.D. degree (or the equivalent) from an academic institution

print(wn.synsets('physician'))
# outpus: [Synset('doctor.n.01')] => relationship = doctor & physician belong to the same synset => synonyms


common_synsets_kid_child = set(wn.synsets('kid', pos=wn.NOUN)).intersection(wn.synsets('child', pos=wn.NOUN))
print(common_synsets_kid_child)
# # output: {Synset('child.n.01'), Synset('child.n.02')} => relationship = synonym

for syn in common_synsets_kid_child:
    print(syn.definition())
# # output: a young person of either sex
# # a human offspring (son or daughter) of any age
```

## e)
* The sentence pair **S1-S2** is sematically closer than *S1-S3*.
* Neither the set representation, nor the bag-of-words vector representation capture this correctly, since both representations report a 100% similarity score for all the metrics used at point *b)* for the pair S1-S3, and a lower score for the pair S1-S2.
* The sentences S1 and S3 appear deceptively similar in a bag-of-words representation because, in such a representation, only the overall (unordered) lexical information is captured. BoW does not take into account the order or the meaning or the words. 
* Sentence embeddings capture the meaning of sentences because they represent words and sentences as dense numerical vectors that encode semantic relationships, word order, and contextual information.
